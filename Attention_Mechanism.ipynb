```python
import torch
import torch.nn as nn
import math


# --------------------------------
# 1. Input embeddings
# --------------------------------

x = torch.tensor([
    [1.0, 0.0, 1.0, 0.0],   # I
    [0.0, 1.0, 0.0, 1.0],   # love
    [1.0, 1.0, 0.0, 0.0]    # dogs
])


# --------------------------------
# 2. Create Q, K, V projections
# --------------------------------

W_q = nn.Linear(4, 4, bias=False)
W_k = nn.Linear(4, 4, bias=False)
W_v = nn.Linear(4, 4, bias=False)


# --------------------------------
# 3. Generate Q, K, V
# --------------------------------

Q = W_q(x)
K = W_k(x)
V = W_v(x)


print("Q:")
print(Q)

print("\nK:")
print(K)

print("\nV:")
print(V)


# --------------------------------
# 4. Calculate attention scores
# --------------------------------

scores = Q @ K.T

print("\nRaw Attention Scores:")
print(scores)


# --------------------------------
# 5. Scale scores
# --------------------------------

d_k = K.size(-1)

scores = scores / math.sqrt(d_k)

print("\nScaled Attention Scores:")
print(scores)


# --------------------------------
# 6. Softmax
# --------------------------------

attention_weights = torch.softmax(
    scores,
    dim=-1
)

print("\nAttention Weights:")
print(attention_weights)


# --------------------------------
# 7. Weighted sum of Values
# --------------------------------

output = attention_weights @ V

print("\nAttention Output:")
print(output)
```
