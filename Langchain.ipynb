```python
from google.colab import userdata

from langchain_google_genai import ChatGoogleGenerativeAI

from langchain_core.messages import (
    SystemMessage,
    HumanMessage
)

# API key
api_key = userdata.get("GOOGLE_API_KEY")

# Model
model = ChatGoogleGenerativeAI(
    model="models/gemini-3.6-flash",
    google_api_key=api_key,
    temperature=0
)

# Conversation
messages = [
    SystemMessage(
        content="You are an AI teacher who explains AI concepts simply."
    ),
    HumanMessage(
        content="What is an embedding?"
    )
]

# Invoke model
response = model.invoke(messages)

# Print answer
print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    "Explain {topic} in simple words."
)

chain = prompt | model

result = chain.invoke({
    "topic": "PyTorch"
})

print(result)
```

# output parsers

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in 2 simple sentences."
)

parser =  StrOutputParser()

chain = prompt | model | parser

result = chain.invoke({
    "topic": "PyTorch"
})

print(result)
```

# streamming

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt | model | StrOutputParser()

for chunk in chain.stream({
    "topic": "Attention Mechanism"
}):
    print(chunk, end="")
```

# text splitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """
PyTorch is a deep learning framework.
It is widely used for machine learning.
Transformers use attention mechanisms.
RAG allows LLMs to use external knowledge.
"""

splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)

chunks = splitter.split_text(text)
for i, chunk in enumerate(chunks):
    print(f"\nChunk {i}:")
    print(chunk)
```

# Vector db

```python
!pip install -q faiss-cpu langchain-community
```

```python
from langchain_core.documents import Document

documents = [
    Document(page_content="PyTorch is a deep learning framework."),
    Document(page_content="Transformers use self-attention mechanisms."),
    Document(page_content="RAG allows language models to use external documents."),
    Document(page_content="Python is a popular programming language.")
]
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    documents,
    embeddings
)
```

```python
query = "What framework is used for deep learning?"

results = vectorstore.similarity_search(query, k=2)

for doc in results:
  print(doc.page_content)
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 2}
)
```

```python
results = retriever.invoke("What framework is used for deep learning?")

for doc in results:
  print(doc.page_content)
```

# Retrieval Augmented Generator(RAG)

```python
from langchain_core.documents import Document

documents = [
    Document(
        page_content="PyTorch is a deep learning framework developed by Meta."
    ),
    Document(
        page_content="Transformers use self-attention to process relationships between tokens."
    ),
    Document(
        page_content="RAG retrieves relevant external documents and provides them to an LLM."
    ),
    Document(
        page_content="LangChain is a framework for building applications powered by language models."
    )
]
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = FAISS.from_documents(
    documents,
    embeddings
)
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 2}
)
```

```python
question = "What does RAG work?"

docs = retriever.invoke(question)

for doc in docs:
  print(doc.page_content)
```

# Agents

```python
from langchain_core.tools import tool

@tool
def add(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b


@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    return a * b

tools = [add, multiply]
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=model,
    tools=tools
)
```
