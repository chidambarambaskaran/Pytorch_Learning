```python
# ============================================================
# SIMPLE LINEAR REGRESSION USING PYTORCH
# ------------------------------------------------------------
# Goal:
# Learn the equation y = 2x + 1 from the given dataset.
#
# This example demonstrates:
# 1. Creating tensors
# 2. Building a Linear Model
# 3. Loss Function
# 4. Optimizer
# 5. Forward Pass
# 6. Backpropagation
# 7. Updating Parameters
# ============================================================

import torch
import torch.nn as nn

# ============================================================
# STEP 1 : CREATE THE DATASET
# ============================================================
# Shape -> (Number of Samples, Number of Features)
#
# Here:
# 4 Samples
# 1 Feature (x)
#
# Equation:
# y = 2x + 1
# ============================================================

x = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

y = torch.tensor([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])

# ============================================================
# STEP 2 : BUILD THE MODEL
# ============================================================
# nn.Linear(in_features, out_features)
#
# in_features  = Number of input features
# out_features = Number of outputs
#
# Here:
# x ---> Linear Layer ---> y
#
# Therefore:
# 1 input
# 1 output
#
# Internally PyTorch creates:
#
# weight (requires_grad=True)
# bias   (requires_grad=True)
#
# Forward Equation:
#
# y = weight*x + bias
# ============================================================

model = nn.Linear(1, 1)

# Print randomly initialized parameters
print("=" * 60)
print("Initial Parameters")
print("=" * 60)

print("Weight :", model.weight.item())
print("Bias   :", model.bias.item())

# ============================================================
# STEP 3 : DEFINE LOSS FUNCTION
# ============================================================
#
# MSE = Mean Squared Error
#
# Formula:
#
# ((prediction - actual)^2).mean()
#
# We manually wrote this earlier.
# PyTorch already provides it.
# ============================================================

criterion = nn.MSELoss()

# ============================================================
# STEP 4 : DEFINE OPTIMIZER
# ============================================================
#
# Optimizer updates the parameters.
#
# We manually did:
#
# weight -= learning_rate * weight.grad
#
# bias   -= learning_rate * bias.grad
#
# optimizer.step() does this automatically.
#
# model.parameters() returns:
#
# weight
# bias
#
# ============================================================

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

# ============================================================
# STEP 5 : TRAINING LOOP
# ============================================================
#
# Every Epoch:
#
# 1. Clear old gradients
# 2. Forward Pass
# 3. Compute Loss
# 4. Backpropagation
# 5. Update Parameters
#
# ============================================================

epochs = 100

for epoch in range(epochs):

    # --------------------------------------------------------
    # Remove previous gradients
    #
    # PyTorch accumulates gradients.
    #
    # If we don't clear them,
    # gradients from previous epochs will be added.
    # --------------------------------------------------------

    optimizer.zero_grad()

    # --------------------------------------------------------
    # FORWARD PASS
    #
    # Equivalent to:
    #
    # prediction = weight*x + bias
    # --------------------------------------------------------

    prediction = model(x)

    # --------------------------------------------------------
    # LOSS
    #
    # Compare prediction with actual values.
    # --------------------------------------------------------

    loss = criterion(prediction, y)

    # --------------------------------------------------------
    # BACKPROPAGATION
    #
    # Computes gradients automatically.
    #
    # weight.grad
    # bias.grad
    #
    # are calculated here.
    # --------------------------------------------------------

    loss.backward()

    # --------------------------------------------------------
    # UPDATE PARAMETERS
    #
    # Internally performs:
    #
    # parameter = parameter - lr * gradient
    # --------------------------------------------------------

    optimizer.step()

    # --------------------------------------------------------
    # Print progress every 10 epochs
    # --------------------------------------------------------

    if epoch % 10 == 0:

        print(
            f"Epoch {epoch:3d}"
            f" | Loss = {loss.item():.6f}"
            f" | Weight = {model.weight.item():.4f}"
            f" | Bias = {model.bias.item():.4f}"
        )

# ============================================================
# STEP 6 : FINAL PARAMETERS
# ============================================================

print("\n" + "=" * 60)
print("Training Completed")
print("=" * 60)

print("Final Weight :", model.weight.item())
print("Final Bias   :", model.bias.item())

# ============================================================
# STEP 7 : TEST THE MODEL
# ============================================================
#
# Predict a new value.
#
# Example:
# x = 10
#
# Expected:
# y ≈ 21
# ============================================================

test = torch.tensor([[10.0]])

prediction = model(test)

print("\nPrediction for x = 10")

print("Predicted y =", prediction.item())
```
